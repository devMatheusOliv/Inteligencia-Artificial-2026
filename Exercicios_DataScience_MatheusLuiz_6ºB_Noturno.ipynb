4 - Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.

```python
import numpy as np

numero_aleatorio = np.random.randint(1, 100, 100)
median = np.median(numero_aleatorio)
mean = np.mean(numero_aleatorio)

contagem = {}
for valor in numero_aleatorio:
    if valor in contagem:
        contagem[valor] += 1
    else:
        contagem[valor] = 1

moda = None
maior_contagem = 0
for valor, vezes in contagem.items():
    if vezes > maior_contagem:
        maior_contagem = vezes
        moda = valor

print(f"Os números são {numero_aleatorio}\n a mediana é {median} a média é {mean} a moda é {moda}")
```

7 - Crie dois arrays de números aleatórios e encontre a correlação entre eles.

```python
import numpy as np
array1 = np.random.rand(30)
array2 = np.random.rand(30)
mean1 = np.mean(array1)
mean2 = np.mean(array2)
cov = np.mean((array1 - mean1) * (array2 - mean2))
std1 = np.std(array1)
std2 = np.std(array2)
correlacao = cov / (std1 * std2)
print(correlacao)
```

19 - Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

u = np.random.randint(1, 100, 20)
v = np.random.randint(1, 100, 20)

prod_escalar = np.dot(u, v)
norm_u = np.sqrt(np.sum(u ** 2))
norm_v = np.sqrt(np.sum(v ** 2))
cos_theta = prod_escalar / (norm_u * norm_v)

print(u)
print(v)
print(f"Cosseno do ângulo: {cos_theta}")
```

## 2) Exercícios de SciPy

1 - Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.

```python
from scipy.integrate import quad
import numpy as np
resultado, erro = quad(lambda x: x**2 * np.sin(x), 0, np.pi)
print(f"Resultado da integral: {resultado}")
print(f"Erro estimado: {erro}")
```

2 - Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np
solucao = solve_ivp(lambda t, y: t - y, [0, 10], [1])
print(f"Solução em x=10: {solucao.y[0][-1]}")
```

3 - Encontre as raízes da função f(x) = x⁴ − 3x³ + 2 utilizando SciPy.

```python
from scipy.optimize import root

def funcao(x):
    return x**4 - 3*x**3 + 2

resultado = root(funcao, 2)
print(f"Aproximação de raiz: {resultado.x[0]}")
```

4 - Use a função de otimização de SciPy para maximizar f(x) = −x² + 4x + 1.

```python
from scipy.optimize import minimize
import numpy as np
resultado = minimize(lambda x: -(-x**2 + 4*x + 1), x0=0)
print(f"Máximo encontrado em x = {-resultado.x[0]}")
print(f"Valor máximo: {-resultado.fun}")
```

5 - Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
from scipy.fft import fft
import numpy as np
t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
frequencias = fft(sinal)
print(f"Transformada de Fourier: {frequencias[:10]}...")
```

7 - Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

9 - Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).

```python
from scipy.optimize import curve_fit
import numpy as np

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

parametros, _ = curve_fit(parabola, x, y)
print(f"Parâmetros ajustados [a, b, c]: {parametros}")
```

10 - Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [0, 2] para x e [0, 1] para y.

```python
from scipy.integrate import quad
import numpy as np
def integrala_exterior(x):
    def integrala_interior(y):
        return x * y + x**2
    resultado_interior, _ = quad(integrala_interior, 0, 1)
    return resultado_interior
resultado_exterior, erro_exterior = quad(integrala_exterior, 0, 2)
print(f"Resultado da integral dupla: {resultado_exterior}")
print(f"Erro estimado: {erro_exterior}")
```

10 - Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y1 = x
y2 = x**2
y3 = x**3
plt.plot(x, y1, color='blue', linestyle='-')
plt.plot(x, y2, color='red', linestyle='--')
plt.plot(x, y3, color='green', linestyle=':')
plt.show()
```

11 - Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.

```python
import numpy as np
import matplotlib.pyplot as plt
categorias = ['X', 'Y', 'Z']
grupo1 = [3, 5, 2]
grupo2 = [2, 3, 4]
grupo3 = [1, 2, 3]
plt.bar(categorias, grupo1, label='Grupo 1')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2')
plt.bar(categorias, grupo3, bottom=np.array(grupo1)+np.array(grupo2), label='Grupo 3')
plt.legend()
plt.title("Barras empilhadas")
plt.show()
```

13 - Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(x, y, z)
plt.show()
```

18 - Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import numpy as np
import matplotlib.pyplot as plt
categorias = ['A', 'B', 'C', 'D']
valores = [5, -3, 7, -2]
cores = ['green' if v >= 0 else 'red' for v in valores]
plt.bar(categorias, valores, color=cores)
plt.show()
```

19 - Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1, label='Seno')
plt.plot(x, y2, label='Cosseno')
plt.legend()
plt.show()
```
