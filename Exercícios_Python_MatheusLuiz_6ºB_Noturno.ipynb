18 - Número Primo - testo divisores ate a raiz quadrada,

```python
def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def verificar_primo():
    n = int(input("Digite um número: "))
    print("É primo" if eh_primo(n) else "Não é primo")

verificar_primo()
```

21 - Busca Binária - Vi isso na matéria de estruturas de dados, divide o intervalo pela metade

```python
def busca_binaria(lista, alvo):
    inicio, fim = 0, len(lista) - 1
    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1

lista_teste = [1, 3, 5, 7, 9, 11, 13]
indice_encontrado = busca_binaria(lista_teste, 9)
print(f"Busca binária: o número 9 está no índice {indice_encontrado}")
```

22 - Criptografia de Cifra de César - Desloco cada letra n posições no alfabeto usando ord() e chr()

```python
def cifra_cesar(texto, deslocamento):
    resultado = ""
    for char in texto:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            resultado += chr((ord(char) - base + deslocamento) % 26 + base)
        else:
            resultado += char
    return resultado
```

23 - Descriptografar Cifra de César - É só chamar a função de cima com o deslocamento negativo

24 - Validação de CPF - Implementei o cálculo dos dois dígitos verificadores conforme a regra oficial

```python
def validar_cpf(cpf):
    cpf = ''.join(filter(str.isdigit, cpf))
    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    def calcular_digito(cpf_parcial):
        soma = 0
        peso = len(cpf_parcial) + 1
        for digito in cpf_parcial:
            soma += int(digito) * peso
            peso -= 1
        resto = soma % 11
        return '0' if resto < 2 else str(11 - resto)

    digito1 = calcular_digito(cpf[:9])
    digito2 = calcular_digito(cpf[:9] + digito1)
    return cpf[-2:] == digito1 + digito2
```

31 - Ordenação por Bolha - compara vizinhos e troca se estiver fora de ordem

```python
def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista
```

46 - Calcular Média e Desvio Padrão - Fiz a fórmula na mão pra treinar

```python
def media_e_desvio_padrao():
    numeros = [10, 12, 23, 23, 16, 23, 21, 16]
    media = sum(numeros) / len(numeros)
    variancia = sum((x - media) ** 2 for x in numeros) / len(numeros)
    desvio_padrao = variancia ** 0.5
    print(f"Média: {media:.2f} | Desvio padrão: {desvio_padrao:.2f}")

media_e_desvio_padrao()
```

48 - Ordenação por Inserção- pego cada elemento e insiro na posição certa

```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = chave
    return lista

print(f"Insertion sort: {insertion_sort([9, 4, 7, 1, 3, 8])}")
```
